```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
sys.path.insert(0, str(ROOT / "scripts"))

import make_eda_figures as eda  # noqa: E402
from aerosentinel.data.ncmapss import UnitStore  # noqa: E402
from aerosentinel.utils import get_paths  # noqa: E402
from aerosentinel.viz import apply_style  # noqa: E402

apply_style()
paths = get_paths()
dev = UnitStore(paths.data_processed / "ncmapss", "DS02", "dev")
dev.units, dev.var_names
```
